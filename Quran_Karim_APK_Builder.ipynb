# ساخت APK برنامه قرآن کریم با گوشی اندرویدی

این دفترچه در Google Colab اجرا می‌شود. ابتدا فایل ZIP پروژه را از گوشی بارگذاری کن؛ سپس سلول‌ها را به‌ترتیب اجرا کن. در پایان، اگر ساخت موفق باشد، فایل APK برای دانلود نمایش داده می‌شود.

**نکته:** اولین ساخت ممکن است زمان‌بر باشد، چون ابزارهای Android SDK و NDK دانلود می‌شوند.

```python
from google.colab import files
uploaded = files.upload()  # فایل quran_kivy_project_with_audio.zip را انتخاب کن
zip_name = next(iter(uploaded))
print('فایل دریافت شد:', zip_name)
```

```python
import os, zipfile, shutil
project_dir = '/content/quran_project'
shutil.rmtree(project_dir, ignore_errors=True)
os.makedirs(project_dir, exist_ok=True)
with zipfile.ZipFile(zip_name) as z:
    z.extractall(project_dir)
print('فایل‌های پروژه:', os.listdir(project_dir))
assert os.path.exists(os.path.join(project_dir, 'quran_app.py'))
assert os.path.exists(os.path.join(project_dir, 'fatiha.mp3'))
assert os.path.exists(os.path.join(project_dir, 'ikhlas.mp3'))
os.rename(os.path.join(project_dir, 'quran_app.py'), os.path.join(project_dir, 'main.py'))
print('آماده‌سازی پروژه انجام شد.')
```

```python
spec = '''[app]
title = Quran Karim
package.name = qurankarim
package.domain = org.amirabbas
source.dir = .
source.include_exts = py,png,jpg,kv,atlas,mp3,txt
version = 1.0
requirements = python3,kivy,arabic-reshaper,python-bidi
orientation = portrait
fullscreen = 0
android.archs = arm64-v8a, armeabi-v7a
android.minapi = 23
android.accept_sdk_license = True
android.permissions =
p4a.bootstrap = sdl2
log_level = 2

[buildozer]
warn_on_root = 0
'''
with open(os.path.join(project_dir, 'buildozer.spec'), 'w', encoding='utf-8') as f:
    f.write(spec)
print(open(os.path.join(project_dir, 'buildozer.spec'), encoding='utf-8').read())
```
